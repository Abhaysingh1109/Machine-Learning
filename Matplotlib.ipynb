```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
```

Matplotlib Basics


```python
x = np.linspace(0,5,11)
x
y = x ** 2
y
```

```python
plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Plot')

```

Subplot in Matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
low = np.linspace(0, 2*np.pi, 100)
high = low ** 3
plt.plot(low,high)
plt.subplot(2,2,3)
low = np.linspace(0, 2*np.pi, 100)
high = low/3
plt.plot(low,high)
plt.subplot(2,2,4)
low = np.linspace(3, 6*np.pi, 230)
high = low**50
plt.plot(low,high,'r*')
```

Object Oriented Way

```python
fig = plt.figure()
axes = fig.add_axes([0.1,0.1,1,1])
axes.plot(x,y)
axes1 = fig.add_axes([0.3,0.3,0.2,0.2])
axes1.plot(x,y)
plt.savefig("figure")
```

TYPE OF PLOTS

```python
plt.scatter(low,high)
```

```python
plt.barh(low,high)
```

```python
plt.fill_between(low,high)
```

```python
data = np.random.randn(1000)

# Create histogram
plt.hist(data, bins=20)

# Add labels
plt.title("Histogram Example")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
plt.savefig('Histogram')
```

```python
plt.boxplot([x,y])
```

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```
