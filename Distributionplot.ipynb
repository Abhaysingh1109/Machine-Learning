```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
df
```

```python
df['size'].unique()
```

```python
sns.histplot(df['total_bill'],bins=20,color="pink")
```

```python

plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins = 20,kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins = 20,kde = True)
```

```python
sns.jointplot(x='total_bill',y = 'tip',data=df,kind='scatter',color='green')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'reg')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'resid',color= 'black')
```

```python
sns.pairplot(df,hue='sex')
```

```python
sns.rugplot(df,palette='rainbow')
```
