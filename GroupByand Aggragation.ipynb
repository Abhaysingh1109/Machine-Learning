Group By

```python
data = {
    'Category': ['A', 'B', "A" , "B", "A", "B", "A", "B"],
    'Store': ['S1', 'S1', "S2" ,"S2", "S1", "S2","S2", "S1"],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 251],
    'Date': pd. date_range('2023-01-01', periods =  8)
}

df = pd.DataFrame(data)
    
        
```

```python
df
```

```python
#GroupBy catergory and calcualte the sum of sales

df.groupby('Category')['Sales'].sum()
```

```python
#GroupBy stores and calculate the sum of sales

df.groupby('Store')['Sales'].sum()
```

```python
# GroupBy multiple column
# GroupBy and category and store
```

```python
cat = df.groupby(['Category','Store'])['Sales'].sum()
cat
```
